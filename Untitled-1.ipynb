```python
import pandas as pd

path = r"D:\Học POWER BI\Ecommerce_Project 1"

Customers = pd.read_csv(path + r"\Customers.csv")
Marketing_Funnel = pd.read_csv(path + r"\Marketing_Funnel.csv")
Products = pd.read_csv(path + r"\Products.csv")
Sales = pd.read_csv(path + r"\Sales.csv")
```

# Kiểm tra Data

```python
# Kiểm tra số dòng và số cột
print("Customers:", Customers.shape)
print("Marketing_Funnel:", Marketing_Funnel.shape)
print("Products:", Products.shape)
print("Sales:", Sales.shape)
```

```python
# Kiểm tra cấu trúc dữ liệu
Customers.info()
Marketing_Funnel.info()
Products.info()
Sales.info()
```

```python
# Kiểm tra giá trị null
for name, df in {
    "Customers": Customers,
    "Marketing_Funnel": Marketing_Funnel,
    "Products": Products,
    "Sales": Sales
}.items():
    print(f"\n--- {name}: Missing Values ---")
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Kiểm tra giá tri trùng lặp
for name, df in {
    "Customers": Customers,
    "Marketing_Funnel": Marketing_Funnel,
    "Products": Products,
    "Sales": Sales
}.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

# Đổi data type

```python
# Customers
Customers["Join_Date"] = pd.to_datetime(
    Customers["Join_Date"], errors="coerce"
)

# Marketing Funnel
Marketing_Funnel["Week_Start"] = pd.to_datetime(
    Marketing_Funnel["Week_Start"], errors="coerce"
)

# Products
Products["Launch_Date"] = pd.to_datetime(
    Products["Launch_Date"], errors="coerce"
)

# Sales
for col in ["Order_Date", "Promised_Date", "Delivered_Date"]:
    Sales[col] = pd.to_datetime(Sales[col], errors="coerce")
```

# Xuất file 

```python
tables = {
    "Customers": Customers,
    "Marketing_Funnel": Marketing_Funnel,
    "Products": Products,
    "Sales": Sales
}

for name, df in tables.items():
    df.to_csv(path + f"\\{name}_cleaned.csv", index=False, encoding="utf-8-sig")
```

```python
# 1. Calculate Revenue
Sales["Revenue"] = (
    Sales["Quantity"] * Sales["Unit_Price"]
    * (1 - Sales["Discount_Pct"])
)

# 2. Descriptive Statistics - Numerical Variables
cols = [
    "Quantity", "Unit_Price", "Discount_Pct",
    "Unit_Cost", "Revenue", "Delivery_Days",
    "Customer_Rating"
]

stats = Sales[cols].describe().T
stats["Variance"] = Sales[cols].var()
stats["Skewness"] = Sales[cols].skew()

print("===== Numerical Descriptive Statistics =====")
display(stats.round(2))

# 3. Age Group Distribution
age_stats = Customers["Age_Group"].value_counts().to_frame("Count")
age_stats["Percentage (%)"] = (
    Customers["Age_Group"].value_counts(normalize=True) * 100
).round(2)

print("===== Age Group Descriptive Statistics =====")
display(age_stats)
```
